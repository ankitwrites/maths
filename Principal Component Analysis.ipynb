```python
#PCA uses the covariant matrix and its eigenvectors to reduce a dataset's dimensionality while retaining its variation

import numpy as np

#five observation with three features
X = np.array([[2, 0, 1], [0, 1, 0], [3, 1, 2], [1, 2, 1], [4, 3, 3]])

print("X =\n {}".format(X))
print("Shape of X = {}".format(X.shape))
```

```python
#Mean of each feature

mean = np.mean(X, axis=0)
print("Mean of each feature = {}".format(mean))

#X_centered = X - mean
X_c = X - mean
print("X_centered =\n {}".format(X_c))

#Covariance Matrix = (X_c^T X_c) / (n - 1)
C = (X_c.T @ X_c) / (X.shape[0] - 1)
print("Covariance Matrix =\n {}".format(C))
```

```python
#Eigenvalues and Eigenvectors of covariance matrix C

eigenvalues, eigenvectors = np.linalg.eigh(C)

#sort eigenvalues from largest to smallest
sort = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[sort]
eigenvectors = eigenvectors[:, sort]

print("Eigenvalues = {}".format(eigenvalues))
print("Eigenvectors =\n {}".format(eigenvectors))
```

```python
#Keep two eigenvectors with the largest eigenvalues because their eigenvectors preserve the greatest amount of variance in the data

W = eigenvectors[:, :2]
print("W =\n {}".format(W))
print("Shape of W = {}".format(W.shape))
```

```python
#Project the data into two dimensions

#X_reduced = X_c W

X_r = X_c @ W

print("Reduced dataset (X_r) =\n {}".format(X_r))
print("Shape of X_r = {}".format(X_r.shape))

#each observation now has two principal component values instead of three original feature values
```

```python
#Information Retained

variance_ratio = eigenvalues / np.sum(eigenvalues)

print("Variance Ratio = {}".format(variance_ratio))
print("Variance Retained = {}".format(np.sum(variance_ratio[:2]) * 100))

#Variance retained = (lambda 1 + lambda 2) / (lambda 1 + lambda 2 + lambda 3) * 100

lambda12 = 0
for i in range(0, 2):
    lambda12 = lambda12 + eigenvalues[i]

print("Variance Retained = {}".format(lambda12 / np.sum(eigenvalues) * 100))
```
