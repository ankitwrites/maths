```python
#differentiation of f(x) using forward, backward and central differences

#define function

def f(x):
    return x**2

#define steps
h = 1
#differentiation at value
x = 1
```

```python
#forward difference

diff = f(x + h) - f(x)

print("(forward difference) f'(2) = {}".format(diff / h))
```

```python
#backward difference

diff = f(x) - f(x - h)

print("(backward difference) f'(2) = {}".format(diff / h))
```

```python
#central difference

diff = f(x + h) - f(x - h)

print("(central difference) f'(2) = {}".format(diff / 2 / h ))
```
