```python
#Determine Eigenvalues

import numpy as np

A = np.array([[2, 0, -1], [0, 2, 0], [-1, 0, 2]])

print("Given matrix: A =\n {}".format(A))
```

```python
#using numpy library

eigenvalues = np.linalg.eigvals(A)

print("Eigenvalues: {}".format(np.round(eigenvalues, 10)))
```

```python
#QR Algorithm

def qrDecom(A):
    Q = np.zeros((A.shape[0], A.shape[0]))
    R = np.zeros((A.shape[0], A.shape[0]))
    v = np.zeros((A.shape[0]))
    for j in range(0, A.shape[0]):
        v = A[:, j]
        for i in range(0, j-1):
            R[i, j] = Q[:, i] @ A[:, j]
            v = v - R[i, j] * Q[:, i]
        R[j, j] = np.sqrt(v @ v)
        Q[:, j] = v / R[j, j]
    return Q, R

for itr in range(0, 100):
    Q, R = qrDecom(A)
    A = R @ Q

print("Eigen Values are:")
for i in range(0, A.shape[0]):
    print(np.round(A[i, i], 10))
```

```python
#QR decomposition using Numpy

A = np.array([[2, 0, -1], [0, 2, 0], [-1, 0, 2]])

for itr in range(0, 100):
    Q, R = np.linalg.qr(A)
    A = R @ Q

print("Eigen Values are:")
for i in range(0, A.shape[0]):
    print(np.round(A[i, i], 10))
```
