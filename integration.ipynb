```python
#integration using Rectangle rule, trapezoidal rule and Simpson's rule

#f(x) = e^(-x^2)

import numpy as np

def f(x):
    return np.exp(-x**2)

#lower limit
a = 0
#upper limit
b = 1
#subdivisons
n = 5
h = (b-a)/n
```

```python
#Rectangle Rule

sum = 0

for i in range(0, n):
    mid = (a + (a + h))/2
    sum = sum + f(mid)
    a = a + h

print("from rectangle rule: I = {}".format(h * sum))
```

```python
#Trapezoidal Rule

a = 0
sum = 0
for i in range(0, n + 1):
    if (i == 0 or i == n):
        sum = sum + f(a)
    else:
        sum = sum + 2 * f(a)
    a = a + h

print("from trapezoidal rule: I = {}".format(h * sum / 2))
```

```python
#Simpson's 1/3 rule

a = 0
sum = 0
n = 4 #requires even number of subintervals
h = (b-a)/n
for i in range(0, n+1):
    if (i == 0 or i == n):
        sum = sum + f(a)
    else:
        if (i%2 == 0):
            sum = sum + 2 * f(a)
        else:
            sum = sum + 4 * f(a)
    a = a + h

print("from Simpson's 1/3 rule: I = {}".format(h * sum / 3))
```

```python
#Simpson's 3/8 rule

a = 0
sum = 0
n = 6 #requires subintervals in the multiple of 3
h = (b-a)/n
for i in range(0, n+1):
    if(i == 0 or i == n):
        sum = sum + f(a)
    else:
        if (i%3 == 0):
            sum = sum + 2 * f(a)
        else:
            sum = sum + 3 * f(a)
    a = a + h

print("from Simpson's 3/8 rule: I = {}".format(3 * h * sum / 8))
```
