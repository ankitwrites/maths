```python
def f(x, y):
    return x + y

#initial values, y(0) = 1
x0 = 0
y0 = 1

#find the value of y at x = 0.6
x1 = 0.6
#let us take a step size
h = 0.005
```

```python
#Euler Method

x = x0
y = y0
while x <= x1:
    y1 = y + h * f(x, y)
    x = x + h
    y = y1

print("From Euler method, y({}) = {}".format(x1, y1))
```

```python
#Midpoint Method

#define k1 and k2

x = x0
y = y0
while x <= x1:
    k1 = f(x, y)
    k2 = f(x + h/2, y + h * k1 / 2)
    y1 = y + h * k2
    x = x + h
    y = y1

print("From Midpoint method, y({}) = {}".format(x1, y1))
```

```python
#RK4 Method

x = x0
y = y0

while x <= x1:
    k1 = f(x, y)
    k2 = f(x + h/2, y + h * k1 / 2)
    k3 = f(x + h/2, y + h * k2 / 2)
    k4 = f(x + h, y + h * k3)
    y1 = y + h * (k1 + 2 * k2 + 2 * k3 + k4) / 6
    x = x + h
    y = y1

print("From RK4 method, y({}) = {}".format(x1, y1))
```
